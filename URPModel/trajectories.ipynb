```python
%cd /home/smalani/PartialObservations/
from URPModel import datagen
from config import config
import matplotlib.pyplot as plt
import numpy as np
```

```python
data_train = datagen.generate_data(n_train=100, x1_sample_num=1000, x2_sample_num=1000, reg_time=True, init_available=True, full_observations=True)
```

```python
Da_sample_array = np.array([0.2,0.25,0.3,0.35,0.4,0.47,0.5])
data_train_Da = []
for Da_sample in Da_sample_array:
    data_train_Da.append(datagen.generate_data(n_train=1, x1_sample_num=1000, x2_sample_num=1000, reg_time=True, init_available=True, full_observations=True, Da_set=Da_sample))
```

```python
fig = plt.figure(figsize=(15,5))
ax1 = fig.add_subplot(121)
ax2 = fig.add_subplot(122)



num_plot = 10
labels = ['x', 'y', 'z', 'u', 'v', 'g']

for j in range(data_train[0].shape[0]):
    ax1.plot(data_train[0][j][0][1], data_train[0][j][0][0], alpha=0.1, c='k')
    ax2.plot(data_train[0][j][1][1], data_train[0][j][1][0], alpha=0.1, c='k')

for j in range(len(data_train_Da)):
    ax1.plot(data_train_Da[j][0][0][0][1], data_train_Da[j][0][0][0][0], linewidth=2, label=str(Da_sample_array[j]))
    ax2.plot(data_train_Da[j][0][0][1][1], data_train_Da[j][0][0][1][0], linewidth=2, label=str(Da_sample_array[j]))

ax1.set_ylabel(r'$x_1$', fontsize=20)
ax2.set_ylabel(r'$x_2$', fontsize=20)
ax1.set_xlabel(r'$t$', fontsize=20)
ax2.set_xlabel(r'$t$', fontsize=20)
ax1.tick_params(labelsize=20, pad=10, length=10, width=2, direction='in')
ax2.tick_params(labelsize=20, pad=10, length=10, width=2, direction='in')
ax1.legend(fontsize=15, bbox_to_anchor=(2, -0.2), ncol=7, frameon=False)
[x.set_linewidth(1.5) for x in ax1.spines.values()]
[x.set_linewidth(1.5) for x in ax2.spines.values()]

fig.suptitle('Trajectories at different Da values', fontsize=20)
plt.text(0.95, -0.2, 'Da values', fontsize=20, transform=ax1.transAxes)
```
